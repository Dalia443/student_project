```python
from students import generate_students
from analysis import calculate_student_metrics, get_class_statistics, save_report

# 1. Génération des 10 étudiants
raw_students = generate_students(10)

# 2. Calcul des âges et des moyennes individuelles
students_data = calculate_student_metrics(raw_students)

# Affichage des données de chaque étudiant
print("--- DONNÉES DES ÉTUDIANTS ---")
for s in students_data:
    print(f"ID: {s['id']} | Naissance: {s['birth_date']} | Âge: {s['age']} ans | Notes: {s['grades']} | Moyenne: {s['average_grade']}")

# 3. Calcul des statistiques globales de la classe
class_stats = get_class_statistics(students_data)

print("\n--- STATISTIQUES DE LA CLASSE ---")
for key, value in class_stats.items():
    print(f"{key}: {value}")

# 4. Sauvegarde automatique dans le dossier results/report.txt
save_report(class_stats)
```
